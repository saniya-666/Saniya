```python
contacts = {
    "Moana" : "0123456789",
    "Sofia" : "0987654321",
    "Meera" : "1234987605"
}
contacts["Moana"] = "0123456789"
contacts["Rohan"] = "9000087654"
existing_contact = contacts.get("Meera","contact not found")
missing_contact = contacts.get("Rahul","contact not found")
print("Safe LookupResults:")
print("Meera:", existing_contact)
print("Rahul:", missing_contact)
print("\ncontact List:")
for name, phone in contacts.items():
    print(f"contact: {name} | phone: {phone}")
```
