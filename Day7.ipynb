```python
import numpy as np
np.random.seed(42)
scores = np.random.randint(50,101,size=(5,3))
mean_scores =  scores.mean(axis=0)
centered_scores = scores - mean_scores
print("Original Scores:\n",scores)
print("\nSubject-wise Mean:\n",mean_scores)
print("\nCentered Scores (After Mean Subtraction):\n",centered_scores)
```

```python
def mean(array):
    n = len(array)
    return np.sum(array)/n
```

```python
mean(np.array([1,2,3,4,5,6]))
```

```python
def variance(array):
    n = len(array)
    m = mean(array)
    return np.sum((array - m)**2)/n
```

```python
variance(np.array([1,2,3,4,5,6]))
```

```python
import numpy as np
data = np.array([10,20,30,40,50])
min_val = data.min()
max_value = data.max()
scaled = (data - min_val) / (max_val - min_val)
print(scaled)
```

```python
import numpy as np
data = np.array([10,20,30,40,50])
mean = data.mean()
std = data.std()
scaled = (data - mean) / std
print(scaled)
```

```python
import numpy as np
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
x= np.array([-2,-1,0,1,2])
print(sigmoid(x))
```

```python
import numpy as np
def relu(x):
    return np.maximum(0,x)
x = np.array([-3,-1,0,2,4])
print(relu(x))
```
